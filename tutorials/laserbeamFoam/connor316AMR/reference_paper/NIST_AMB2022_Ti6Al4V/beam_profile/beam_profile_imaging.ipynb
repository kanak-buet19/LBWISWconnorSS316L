```python
import numpy as np
import matplotlib.pyplot as plt
import math

LaserPower = 201.3; #Laser power in Watts. (User Input)

#Import the beam profile data whose integral has been normalized to 1. 
#User should input correct file location in the parentheses. 
data = np.genfromtxt("//jake.nist.gov/686/68601/Laser Welding - Shared/Data/AMBenchlet/General Experiment Information/Beam Profile_Normalized Integral.csv", delimiter=',')

Check = np.sum(data*LaserPower); #Check that the integral of the scaled data is approximately the laser power.
                        #Integration rounding errors could exist.
pctDiff = (LaserPower-Check)/LaserPower*100

print(LaserPower, ' = ', round(Check,1), '?')
print("Integration error: ", round(pctDiff,2), ' %')
```

```python
# Generate a contour plot of the scaled beam profile. 

X = np.linspace(0,(5.5*59),num=60) #pixel spacing is 5.5 microns and the file is 60 x 60 pixels.

scale = np.linspace(-.5,math.ceil(np.max(data*LaserPower)),3*(math.ceil(np.max(data*LaserPower))+1));
contour = plt.contourf(X,X,data*LaserPower,scale);
cbar = plt.colorbar(contour);
cbar.set_label('Power (W)', fontsize = 12)
plt.xlim(85,200);
plt.ylim(85,200);
plt.title("Scaled Beam Profile Data")
plt.xlabel('Distance (um)',fontsize=12);
plt.ylabel('Distance (um)', fontsize=12);
```
